# Algoritmo de Entrenamiento en Python

```
import numpy as np

# Función de activación (Función escalón)
def step_function(x):
    return 1 if x >= 0 else 0

# Algoritmo de entrenamiento
def train(X, D, alpha, epochs):
    # Inicializar pesos W y umbral theta aleatoriamente en el rango [-1, 1]
    n_features = X.shape[1]
    W = np.random.uniform(-1, 1, n_features)
    theta = np.random.uniform(-1, 1)
    
    for epoch in range(epochs):
        total_error = 0
        for i in range(len(X)):
            # Paso 3: Propagar
            y_in = np.dot(X[i], W) - theta
            y = step_function(y_in)
            
            # Paso 4: Calcular el error
            error = D[i] - y
            total_error += error ** 2
            
            # Paso 5: Retropropagación
            for j in range(n_features):
                W[j] += alpha * error * X[i][j]
            theta -= alpha * error
            
        # Condición de parada si el error total es aceptable
        if total_error == 0:
            break
    
    return W, theta

# Datos de entrada (X) y salida esperada (D)
X = np.array([
    [1, 0],
    [0, 1],
    [1, 1],
    [0, 0]
])
D = np.array([1, 1, 0, 0])  # Salidas deseadas para el ejemplo

# Parámetros de entrenamiento
alpha = 0.1  # Coeficiente de aprendizaje
epochs = 100  # Número máximo de épocas

# Entrenamiento
W, theta = train(X, D, alpha, epochs)

print("Pesos entrenados:", W)
print("Umbral entrenado:", theta)
```
